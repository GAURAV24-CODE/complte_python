## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
with open('context_demo.txt', 'w', encoding='utf-8') as file:
    file.write('Managed resource')
print('File closed automatically after the with block')
```

### Example 2


```python
from contextlib import contextmanager
@contextmanager
def simple_context():
    print('Enter')
    try:
        yield
    finally:
        print('Exit')
with simple_context():
    print('Inside')
```
