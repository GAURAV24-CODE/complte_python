## Learning goals

- Test normal, boundary and invalid inputs.
- Keep tests repeatable and independent.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import unittest
def add(a, b):
    return a + b
class TestAdd(unittest.TestCase):
    def test_positive_numbers(self):
        self.assertEqual(add(2, 3), 5)
    def test_negative_number(self):
        self.assertEqual(add(-2, 2), 0)
# Run with unittest.main() in a standalone .py file
```
