## Learning goals
- Levels include DEBUG, INFO, WARNING, ERROR and CRITICAL.
- Use logging instead of scattered print statements in applications.
- Include useful context in log messages.
- Debug by reproducing the issue, inspecting state and testing a fix.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
logging.info('Application started')
logging.warning('Input is missing')
```

### Example 2


```python
def calculate_average(values):
    if not values:
        logging.error('Cannot average an empty list')
        return None
    return sum(values) / len(values)
print(calculate_average([]))
```
