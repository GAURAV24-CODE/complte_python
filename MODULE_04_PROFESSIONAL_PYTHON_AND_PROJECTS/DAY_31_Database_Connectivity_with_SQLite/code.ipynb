## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import sqlite3
connection = sqlite3.connect(':memory:')
connection.execute('CREATE TABLE students (id INTEGER, name TEXT)')
connection.execute('INSERT INTO students VALUES (?, ?)', (1, 'Mira'))
rows = connection.execute('SELECT * FROM students').fetchall()
print(rows)
connection.close()
```
