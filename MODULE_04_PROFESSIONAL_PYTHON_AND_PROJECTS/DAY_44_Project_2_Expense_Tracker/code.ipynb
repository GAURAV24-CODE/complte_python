### Example 1


```python
expenses = [{'amount': 120, 'category': 'Travel'}, {'amount': 80, 'category': 'Food'}, {'amount': 50, 'category': 'Food'}]
summary = {}
for item in expenses:
    category = item['category']
    summary[category] = summary.get(category, 0) + item['amount']
print(summary)
```
