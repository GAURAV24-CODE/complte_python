### Example 1


```python
from bs4 import BeautifulSoup
html = '<article><h2>Python Basics</h2><p>Beginner guide</p></article>'
soup = BeautifulSoup(html, 'html.parser')
record = {'title': soup.select_one('h2').get_text(strip=True), 'summary': soup.select_one('p').get_text(strip=True)}
print(record)
```
