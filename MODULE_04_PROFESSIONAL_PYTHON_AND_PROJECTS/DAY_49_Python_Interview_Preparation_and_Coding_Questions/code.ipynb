## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
def first_non_repeating_char(text):
    counts = {}
    for char in text:
        counts[char] = counts.get(char, 0) + 1
    for char in text:
        if counts[char] == 1:
            return char
    return None
print(first_non_repeating_char('swiss'))
```

### Example 2


```python
def is_anagram(first, second):
    return sorted(first.replace(' ', '').lower()) == sorted(second.replace(' ', '').lower())
print(is_anagram('listen', 'silent'))
```
