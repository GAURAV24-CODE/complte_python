## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import sys
items = [1, 2, 3]
print('Reference count (implementation detail):', sys.getrefcount(items))
```

### Example 2


```python
import gc
print('Garbage collection enabled:', gc.isenabled())
print('Collected objects:', gc.collect())
```
