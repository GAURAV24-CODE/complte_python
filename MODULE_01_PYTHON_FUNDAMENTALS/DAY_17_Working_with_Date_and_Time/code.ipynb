## Learning goals
- date represents a calendar date.
- datetime represents date and time.
- timedelta represents a duration.
- strftime formats dates; strptime parses date strings.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
from datetime import date, datetime, timedelta
print(date.today())
print(datetime.now())
```

### Example 2


```python
today = date.today()
next_week = today + timedelta(days=7)
print(next_week)
```

### Example 3


```python
parsed = datetime.strptime('14-11-2026', '%d-%m-%Y')
print(parsed.strftime('%A, %d %B %Y'))
```
