# Day 13 — Lambda, Map, Filter and Reduce
**Module:** Python Fundamentals

## Learning goals
- lambda creates a small anonymous function.
- map applies a function to each item.
- filter keeps items that pass a condition.
- functools.reduce combines items cumulatively.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
double = lambda x: x * 2
print(double(7))
```

### Example 2


```python
numbers = [1, 2, 3, 4, 5]
print(list(map(lambda x: x * 2, numbers)))
print(list(filter(lambda x: x % 2 == 0, numbers)))
```

### Example 3


```python
from functools import reduce
print(reduce(lambda a, b: a + b, [1, 2, 3, 4]))
```
