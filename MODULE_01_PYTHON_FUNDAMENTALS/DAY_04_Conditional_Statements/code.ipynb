# Day 04 — Conditional Statements
**Module:** Python Fundamentals

## Learning goals
- if runs a block when its condition is true.
- elif checks another condition if earlier conditions failed.
- else handles all remaining cases.
- Indentation defines the block.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
score = 76
if score >= 90:
    grade = 'A'
elif score >= 75:
    grade = 'B'
else:
    grade = 'Needs improvement'
print(grade)
```

### Example 2


```python
temperature = 28
if temperature > 30:
    print('Hot')
else:
    print('Comfortable or cool')
```
