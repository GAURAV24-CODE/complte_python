## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
from pathlib import Path
folder = Path('practice_data')
folder.mkdir(exist_ok=True)
file_path = folder / 'notes.txt'
file_path.write_text('Safe path handling', encoding='utf-8')
print(file_path.exists())
```

### Example 2


```python
content = file_path.read_bytes()
print(content)
print(content.decode('utf-8'))
```
