## Learning goals
- open() supports modes such as r, w, a and x.
- read(), readline() and readlines() read content.
- write() writes text; use newline characters for line breaks.
- with manages file closing automatically.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
from pathlib import Path
path = Path('sample.txt')
path.write_text('Python file handling\n', encoding='utf-8')
print(path.read_text(encoding='utf-8'))
```

### Example 2


```python
with open('sample.txt', 'a', encoding='utf-8') as file:
    file.write('Second line\n')
print(Path('sample.txt').read_text(encoding='utf-8'))
```
