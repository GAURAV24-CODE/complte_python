# Day 14 — List, Dictionary and Set Comprehensions
**Module:** Python Fundamentals

## Learning goals
- List comprehension: [expression for item in iterable].
- Add if conditions to filter items.
- Dictionary comprehension creates key-value mappings.
- Set comprehension creates unique values.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
squares = [n ** 2 for n in range(1, 6)]
print(squares)
```

### Example 2


```python
evens = [n for n in range(20) if n % 2 == 0]
print(evens)
```

### Example 3


```python
square_map = {n: n ** 2 for n in range(1, 5)}
unique_lengths = {len(word) for word in ['cat', 'tiger', 'dog']}
print(square_map, unique_lengths)
```
