## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
marks = [78, 85, 91]
marks.append(88)
print(marks)
```

### Example 2


```python
print(marks[0], marks[-1])
print(marks[1:3])
```

### Example 3


```python
numbers = [4, 1, 9, 2]
numbers.sort()
print(numbers)
```
