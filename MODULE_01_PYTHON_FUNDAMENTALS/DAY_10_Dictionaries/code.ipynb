# Day 10 — Dictionaries
**Module:** Python Fundamentals

## Learning goals
- Create dictionaries with {key: value}.
- Keys must be hashable and unique.
- Use get() for safe lookup.
- keys(), values() and items() support iteration.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
student = {'name': 'Mira', 'age': 19, 'course': 'Python'}
print(student['name'])
```

### Example 2


```python
student['city'] = 'Nashik'
student['age'] = 20
print(student.get('email', 'Not provided'))
```

### Example 3


```python
for key, value in student.items():
    print(key, '=>', value)
```
