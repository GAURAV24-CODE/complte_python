## Learning goals
- iter() obtains an iterator; next() requests its next value.
- A generator function uses yield.
- Generator expressions use parentheses.
- Generators are useful for streaming or large data.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
numbers = iter([10, 20, 30])
print(next(numbers))
print(next(numbers))
```

### Example 2


```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1
print(list(count_up_to(5)))
```

### Example 3


```python
squares = (n * n for n in range(5))
print(next(squares))
```
