## Learning goals
- re.search finds a match anywhere in text.
- re.match checks from the beginning; re.fullmatch checks the entire string.
- Character classes and quantifiers describe patterns.
- Use raw strings such as r'\d+' for regex patterns.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import re
text = 'Order IDs: 125, 240 and 981'
print(re.findall(r'\d+', text))
```

### Example 2


```python
email = 'student@example.com'
pattern = r'^[\w.-]+@[\w.-]+\.\w+$'
print(bool(re.fullmatch(pattern, email)))
```

### Example 3


```python
sentence = 'Python   is   useful'
print(re.sub(r'\s+', ' ', sentence))
```
