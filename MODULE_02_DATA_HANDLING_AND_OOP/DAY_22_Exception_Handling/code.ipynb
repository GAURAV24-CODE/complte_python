# Day 22 — Exception Handling
**Module:** Data Handling and OOP

## Learning goals
- try contains code that may raise an exception.
- except handles selected exception types.
- else runs when no exception occurred.
- finally runs whether an exception occurred or not; raise can signal invalid conditions.

Work through the examples, predict outputs before running them, and complete the exercises at the end.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return 'Cannot divide by zero'
print(safe_divide(10, 0))
```

### Example 2


```python
try:
    value = int('42')
except ValueError as error:
    print('Invalid number:', error)
else:
    print('Parsed:', value)
finally:
    print('Finished')
```
