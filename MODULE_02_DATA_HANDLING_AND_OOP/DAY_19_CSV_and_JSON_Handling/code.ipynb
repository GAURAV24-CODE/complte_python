## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
import csv
from pathlib import Path
rows = [{'name': 'Mira', 'score': 91}, {'name': 'Aarav', 'score': 84}]
with open('scores.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['name', 'score'])
    writer.writeheader()
    writer.writerows(rows)
print(Path('scores.csv').read_text(encoding='utf-8'))
```

### Example 2


```python
import json
data = {'project': 'Python', 'days': 50, 'completed': True}
with open('project.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=4)
with open('project.json', encoding='utf-8') as f:
    loaded = json.load(f)
print(loaded)
```
