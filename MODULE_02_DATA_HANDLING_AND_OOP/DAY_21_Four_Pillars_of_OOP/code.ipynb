## Learning goals
- Encapsulation groups data and methods and controls access by convention or properties.
- Abstraction exposes essential behavior while hiding implementation details.
- Inheritance lets a class reuse or extend another class.
- Polymorphism lets different objects respond to the same interface.

## 1. Practical examples
Run each example in its own code cell or copy it into the cells below.

### Example 1


```python
class Animal:
    def speak(self):
        return 'Some sound'
class Dog(Animal):
    def speak(self):
        return 'Woof'
print(Dog().speak())
```

### Example 2


```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    @property
    def celsius(self):
        return self._celsius
print(Temperature(25).celsius)
```
